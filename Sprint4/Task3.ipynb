### Вариант задания № 6

Создать базовый класс Movie в C#, который будет представлять информацию о фильмах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Movie:
• Атрибуты: Название (Title), Режиссер (Director), Год выпуска (ReleaseYear).
• Методы:
  o GetInfo(): метод для получения информации о фильме в виде строки.
  o Watch(): метод для вывода сообщения о просмотре фильма.
  o Rate(): метод для оценки фильма.

Требования к производным классам:
1. Документальный фильм (Documentary): дополнительные атрибуты (Theme). Метод Watch() переопределен для добавления информации о тематике.
2. Игровой фильм (FeatureFilm): дополнительные атрибуты (Genre). Метод Rate() переопределен для логики оценки в зависимости от жанра.
3. Анимационный фильм (AnimatedMovie): дополнительные атрибуты (AnimationStudio). Метод GetInfo() переопределен для включения студии анимации.

```
using System;

// ==========================================
// 1. ИНТЕРФЕЙСЫ
// ==========================================
public interface IPlayable
{
    void Play();
    void Pause();
}

public interface IMediaContent : IPlayable
{
    string Title { get; set; }
    string Director { get; set; }
    int ReleaseYear { get; set; }
    int DurationMinutes { get; set; }
    void Watch();
    void Rate(double score);
    string GetInfo();
}

// ==========================================
// 2. БАЗОВЫЙ КЛАСС MOVIE
// ==========================================
public class Movie : IMediaContent
{
    private string _title;
    private string _director;
    private int _releaseYear;
    private int _durationMinutes;
    private decimal _budget;
    private string _country;
    private int _viewCount;

    public string Title
    {
        get => _title;
        set => _title = string.IsNullOrWhiteSpace(value) ? "Без названия" : value;
    }

    public string Director
    {
        get => _director;
        set => _director = string.IsNullOrWhiteSpace(value) ? "Неизвестный режиссер" : value;
    }

    public int ReleaseYear
    {
        get => _releaseYear;
        set
        {
            if (value > 1888)
                _releaseYear = value;
            else
                throw new ArgumentOutOfRangeException("Год выпуска должен быть позже 1888 года!");
        }
    }

    public int DurationMinutes
    {
        get => _durationMinutes;
        set
        {
            if (value > 0)
                _durationMinutes = value;
            else
                throw new ArgumentOutOfRangeException("Длительность должна быть положительной!");
        }
    }

    public decimal Budget
    {
        get => _budget;
        set
        {
            if (value >= 0)
                _budget = value;
            else
                throw new ArgumentOutOfRangeException("Бюджет не может быть отрицательным!");
        }
    }

    public string Country
    {
        get => _country;
        set => _country = string.IsNullOrWhiteSpace(value) ? "Не указана" : value;
    }

    public int ViewCount => _viewCount;
    public static int TotalMovies { get; private set; } = 0;

    public Movie(string title, string director, int releaseYear, int duration, decimal budget, string country)
    {
        Title = title;
        Director = director;
        ReleaseYear = releaseYear;
        DurationMinutes = duration;
        Budget = budget;
        Country = country;
        _viewCount = 0;
        TotalMovies++;
    }

    // --- Обычные методы класса ---
    public virtual string GetInfo()
    {
        return $"Фильм: «{Title}» | Режиссер: {Director} | Год: {ReleaseYear} | Хронометраж: {DurationMinutes} мин. | Страна: {Country}";
    }

    public virtual void Watch()
    {
        _viewCount++;
        Console.WriteLine($"[Просмотр] Вы начали просмотр фильма «{Title}». Всего просмотров: {_viewCount}.");
    }

    public virtual void Rate(double score)
    {
        Console.WriteLine($"[Оценка] Фильм «{Title}» оценен на {score}/10.");
    }

    // --- ЯВНАЯ РЕАЛИЗАЦИЯ ИНТЕРФЕЙСА IPlayable ---
    void IPlayable.Play()
    {
        _viewCount++;
        Console.WriteLine($"[Явная реализация IPlayable.Play] Запуск медиапотока фильма «{Title}».");
    }

    void IPlayable.Pause()
    {
        Console.WriteLine($"[Явная реализация IPlayable.Pause] Воспроизведение «{Title}» приостановлено.");
    }

    // Дополнительные методы базового класса
    public void CalculateProfit(decimal boxOffice)
    {
        decimal profit = boxOffice - Budget;
        Console.WriteLine($"[Экономика] «{Title}»: Прибыль составляет {profit} млн.$ (Сборы: {boxOffice} млн.$, Бюджет: {Budget} млн.$).");
    }

    public void DisplayAge(int currentYear = 2026)
    {
        Console.WriteLine($"[Возраст] Фильму «{Title}» уже {currentYear - ReleaseYear} лет.");
    }
}

// ==========================================
// 3. ПРОИЗВОДНЫЙ КЛАСС DOCUMENTARY
// ==========================================
public class Documentary : Movie
{
    private string _theme;
    public string Theme
    {
        get => _theme;
        set => _theme = value;
    }

    public bool HasRealFootage { get; set; }
    public string PrimarySource { get; set; }
    public string Narrator { get; set; }

    public Documentary(string title, string director, int releaseYear, int duration, decimal budget, string country,
                       string theme, bool hasRealFootage, string primarySource, string narrator)
        : base(title, director, releaseYear, duration, budget, country)
    {
        Theme = theme;
        HasRealFootage = hasRealFootage;
        PrimarySource = primarySource;
        Narrator = narrator;
    }

    // Перекрытие метода Watch()
    public override void Watch()
    {
        base.Watch();
        Console.WriteLine($"[Документальный акцент] Тематика фильма: «{Theme}». Текст читает: {Narrator}.");
    }

    public void VerifyFacts()
    {
        Console.WriteLine($"[Научная проверка] Источник материала «{Title}»: {PrimarySource}. Использованы архивные съемки: {(HasRealFootage ? "Да" : "Нет")}.");
    }

    public void RecommendForEducation(string faculty)
    {
        Console.WriteLine($"[Образование] Фильм рекомендован для студентов направления: {faculty}.");
    }
}

// ==========================================
// 4. ПРОИЗВОДНЫЙ КЛАСС FEATUREFILM
// ==========================================
public class FeatureFilm : Movie
{
    private string _genre;
    public string Genre
    {
        get => _genre;
        set => _genre = value;
    }

    public string AgeRating { get; set; }
    public string LeadingActor { get; set; }
    public int OscarNominations { get; set; }

    public FeatureFilm(string title, string director, int releaseYear, int duration, decimal budget, string country,
                       string genre, string ageRating, string leadingActor, int nominations)
        : base(title, director, releaseYear, duration, budget, country)
    {
        Genre = genre;
        AgeRating = ageRating;
        LeadingActor = leadingActor;
        OscarNominations = nominations;
    }

    // Перекрытие метода Rate() с жанровой логикой
    public override void Rate(double score)
    {
        if (Genre == "Драма" && score >= 8.5)
        {
            Console.WriteLine($"[Выбор критиков] Игровая драма «{Title}» удостоена высшей награды критиков с оценкой {score}/10!");
        }
        else
        {
            Console.WriteLine($"[Оценка] Игровой фильм «{Title}» ({Genre}, Возрастной рейтинг: {AgeRating}): {score}/10.");
        }
    }

    public void ShowCastInfo()
    {
        Console.WriteLine($"[В главных ролях] В картине «{Title}» главную роль исполнил(а): {LeadingActor}.");
    }

    public void PrintFestivalStatus()
    {
        Console.WriteLine($"[Кинофестиваль] Фильм выдвинут в {OscarNominations} официальных номинациях.");
    }
}

// ==========================================
// 5. ПРОИЗВОДНЫЙ КЛАСС ANIMATEDMOVIE
// ==========================================
public class AnimatedMovie : Movie
{
    private string _animationStudio;
    public string AnimationStudio
    {
        get => _animationStudio;
        set => _animationStudio = value;
    }

    public string AnimationStyle { get; set; }
    public string TargetAudience { get; set; }
    public bool Is3D { get; set; }

    public AnimatedMovie(string title, string director, int releaseYear, int duration, decimal budget, string country,
                         string studio, string style, string targetAudience, bool is3D)
        : base(title, director, releaseYear, duration, budget, country)
    {
        AnimationStudio = studio;
        AnimationStyle = style;
        TargetAudience = targetAudience;
        Is3D = is3D;
    }

    // Перекрытие метода GetInfo()
    public override string GetInfo()
    {
        return $"{base.GetInfo()} | Студия анимации: {AnimationStudio} | Стиль: {AnimationStyle}";
    }

    public void PlaySoundtrack()
    {
        Console.WriteLine($"[Саундтрек] Звучит оригинальная композиция студии {AnimationStudio}.");
    }

    public void CheckAudienceSuitability()
    {
        Console.WriteLine($"[Прокат] Анимация «{Title}» рассчитана на категорию: {TargetAudience}. Формат 3D: {(Is3D ? "Поддерживается" : "2D-версия")}.");
    }
}

// ==========================================
// 6. УПРАВЛЕНИЕ ЗАВИСИМОСТЯМИ (DEPENDENCY INJECTION)
// ==========================================

// Сервис онлайн-кинотеатра (зависит от интерфейса IMediaContent)
public class CinemaStreamingService
{
    private readonly IMediaContent _media;

    // Внедрение зависимости через конструктор
    public CinemaStreamingService(IMediaContent media)
    {
        _media = media;
    }

    public void StartBroadcast()
    {
        Console.WriteLine($"\n[Онлайн-кинотеатр] Запуск эфира: {_media.Title} ({_media.ReleaseYear} г.).");

        // Доступ к методу явной реализации интерфейса через ссылку
        _media.Play();
        _media.Watch();
        Console.WriteLine($"[Онлайн-кинотеатр] Эфир «{_media.Title}» транслируется в отличном качестве.");
    }

    public void StopBroadcast()
    {
        _media.Pause();
        Console.WriteLine($"[Онлайн-кинотеатр] Эфир «{_media.Title}» завершен.");
    }
}

// Сервис кинокритики (зависит от интерфейса IMediaContent)
public class FilmReviewService
{
    private readonly IMediaContent _film;

    // Внедрение зависимости через конструктор
    public FilmReviewService(IMediaContent film)
    {
        _film = film;
    }

    public void PublishCritique(double score, string reviewNotes)
    {
        Console.WriteLine($"\n[Пресс-центр] Публикация рецензии на: {_film.Title}");
        Console.WriteLine(_film.GetInfo());
        _film.Rate(score);
        Console.WriteLine($"[Мнение эксперта]: \"{reviewNotes}\"");
    }
}

// ==========================================
// 7. ДЕМОНСТРАЦИЯ РАБОТЫ ВСЕХ КОМПОНЕНТОВ
// ==========================================
Documentary doc = new Documentary(
    "Планета Земля", "Аластер Фозергилл", 2006, 500, 25m, "Великобритания",
    "Дикая природа", true, "Архивы BBC", "Дэвид Аттенборо"
);

FeatureFilm feature = new FeatureFilm(
    "Оппенгеймер", "Кристофер Нолан", 2023, 180, 100m, "США",
    "Драма", "18+", "Киллиан Мерфи", 13
);

AnimatedMovie anime = new AnimatedMovie(
    "Унесённые призраками", "Хаяо Миядзаки", 2001, 125, 19m, "Япония",
    "Studio Ghibli", "Рисованная 2D", "Для всей семьи", false
);

// 1. Полиморфизм и дополнительные методы
Console.WriteLine("=== 1. БАЗОВАЯ ИНФОРМАЦИЯ И ПОЛИМОРФИЗМ ===");
Console.WriteLine(doc.GetInfo());
doc.Watch();
doc.VerifyFacts();
doc.RecommendForEducation("География");

Console.WriteLine();
Console.WriteLine(feature.GetInfo());
feature.Watch();
feature.Rate(9.3); // Сработает логика драмы
feature.ShowCastInfo();
feature.PrintFestivalStatus();
feature.CalculateProfit(957m);

Console.WriteLine();
Console.WriteLine(anime.GetInfo());
anime.Watch();
anime.PlaySoundtrack();
anime.CheckAudienceSuitability();

// 2. Демонстрация явной реализации интерфейса
Console.WriteLine("\n=== 2. ЯВНАЯ РЕАЛИЗАЦИЯ ИНТЕРФЕЙСА (IPlayable) ===");
// feature.Play(); // Ошибка компиляции! Метод скрыт из публичного интерфейса класса.
// Вызов возможен только через интерфейсную ссылку:
IPlayable playableMovie = feature;
playableMovie.Play();
playableMovie.Pause();

// 3. Управление зависимостями (Dependency Injection)
Console.WriteLine("\n=== 3. УПРАВЛЕНИЕ ЗАВИСИМОСТЯМИ (DEPENDENCY INJECTION) ===");
CinemaStreamingService docBroadcast = new CinemaStreamingService(doc);
docBroadcast.StartBroadcast();
docBroadcast.StopBroadcast();

CinemaStreamingService animeBroadcast = new CinemaStreamingService(anime);
animeBroadcast.StartBroadcast();
animeBroadcast.StopBroadcast();

FilmReviewService reviewService = new FilmReviewService(feature);
reviewService.PublishCritique(9.5, "Монументальная биографическая драма с выдающимся звуковым дизайном.");

Console.WriteLine($"\n[Итог] Всего создано фильмов в каталоге: {Movie.TotalMovies}");
```
