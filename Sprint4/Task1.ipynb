<h1 style="color:DodgerBlue">Явная реализация интерфейса</h1>

В C# интерфейсы могут быть реализованы явным и неявным образом. Давайте подробно рассмотрим, что такое явная реализация интерфейсов, реализация интерфейсов в базовых и производных классах, а также наследование интерфейсов, используя примеры классов `Vehicle`, `Car` и `Motorcycle`.

### Явная реализация интерфейсов

Явная реализация интерфейса используется для того, чтобы скрыть методы интерфейса от стандартного контекста класса. Это означает, что методы интерфейса могут быть вызваны только через ссылку на интерфейс, а не через экземпляр класса.

#### Пример явной реализации

Рассмотрим интерфейс `IVehicle` и его явную реализацию в классе `Car`.

```csharp
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

public class Car : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    private int currentSpeed;

    // Явная реализация интерфейса
    void IVehicle.Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    void IVehicle.Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }

    // Неявная реализация метода
    public int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

В этом примере методы `Start` и `Stop` реализованы явно, и для их вызова необходимо использовать объект типа `IVehicle`:

```csharp

IVehicle myCar = new Car { Make = "Toyota", Model = "Camry" };
myCar.Start(); // Корректно, вызывает явную реализацию
// myCar.Stop(); // Корректно, вызывает явную реализацию
```

Если вы попытаетесь вызвать `Start` или `Stop` непосредственно через экземпляр `Car`, это приведёт к ошибке компиляции:

```csharp
Car myCar = new Car();
myCar.Start(); // Ошибка компиляции
```

### Реализация интерфейсов в базовых и производных классах

Интерфейсы могут быть реализованы в базовых классах и наследоваться в производных классах. Давайте рассмотрим эту концепцию на примере, включающем классы `Vehicle`, `Car` и `Motorcycle`.

#### Базовый класс Vehicle

```csharp
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public abstract void Start();  // Оставляем абстрактным
    public abstract void Stop();   // Оставляем абстрактным

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

#### Реализация в производных классах

Теперь реализуем `Car` и `Motorcycle`.

```csharp
public class Car : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }
}

public class Motorcycle : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }
}
```

Теперь оба класса `Car` и `Motorcycle` обязаны реализовать методы `Start` и `Stop`, которые определены в абстрактном классе `Vehicle`.

### Наследование интерфейсов

Интерфейсы могут наследовать друг от друга. Это позволяет создавать сложные иерархии интерфейсов. Например, мы можем создать интерфейс `IMotorizedVehicle`, который наследует `IVehicle`.

```csharp
public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
}

public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}
```

Таким образом, `Motorcycle` теперь может реализовать все методы из обоих интерфейсов `IVehicle` и `IMotorizedVehicle`.

### Заключение

1. **Явная реализация** интерфейсов скрывает функциональность классов от общего доступа, что помогает избежать путаницы с методами.
   
2. **Реализация интерфейсов в базовых и производных классах** позволяет создавать иерархию классов и обеспечивает обязательность реализации методов в производных классах.

3. **Наследование интерфейсов** позволяет комбинировать несколько интерфейсов, создавая более сложные структуры, что делает код более гибким и расширяемым.

Эти механизмы являются важными инструментами в инструменте разработчика, позволяя создавать более структурированные и управляемые программы.

<h4 style="color:DodgerBlue">Для проверки напишите пример кода на основе классов Vehicle, Car и Motorcycle ниже в блоке с применением интерфейса и абстактного класса:</h4>

````
using System;

// 1. Базовый интерфейс
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

// 2. Наследование интерфейсов
public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
}

// 3. Базовый абстрактный класс
public abstract class Vehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}

// 4. Производный класс Car с ЯВНОЙ реализацией интерфейса IVehicle
public class Car : Vehicle, IVehicle
{
    public int NumberOfDoors { get; set; }

    // Явная реализация методов интерфейса IVehicle (без модификатора public)
    void IVehicle.Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"[Явная реализация] Автомобиль {Make} {Model} (Дверей: {NumberOfDoors}) запущен.");
    }

    void IVehicle.Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"[Явная реализация] Автомобиль {Make} {Model} остановлен.");
    }

    // Обычный метод класса
    public void Accelerate(int amount)
    {
        currentSpeed += amount;
        Console.WriteLine($"Автомобиль {Make} {Model} едет со скоростью {currentSpeed} км/ч.");
    }
}

// 5. Производный класс Motorcycle, реализующий расширенный интерфейс
public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public bool HasSidecar { get; set; }

    public void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"Мотоцикл {Make} {Model} запущен.");
    }

    public void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"Мотоцикл {Make} {Model} остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine($"Мотоцикл {Make} {Model} громко газует: Вруум-вруум!");
    }
}

// --- Проверка работы ---
Console.WriteLine("=== 1. Проверка явной реализации интерфейса (Car) ===");
Car carObj = new Car { Make = "Toyota", Model = "Camry", NumberOfDoors = 4 };

// Прямой вызов carObj.Start() вызовет ошибку компиляции.
// Вызов доступен только через приведение к интерфейсу IVehicle:
IVehicle carAsVehicle = carObj;
carAsVehicle.Start();
carObj.Accelerate(60);
Console.WriteLine($"Скорость автомобиля: {carObj.GetCurrentSpeed()} км/ч");
carAsVehicle.Stop();

Console.WriteLine("\n=== 2. Проверка наследования интерфейсов (Motorcycle) ===");
Motorcycle motoObj = new Motorcycle { Make = "Harley-Davidson", Model = "Street 750", HasSidecar = false };
motoObj.Start();
motoObj.RevEngine();
motoObj.Stop();
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также применить явнаю реализацию интерфейса.

````
using System;
using System.Collections.Generic;

// 1. Базовый интерфейс животного
public interface IAnimal
{
    void MakeSound();
    void Sleep(int hours);
    int GetEnergy();
}

// 2. Расширенный интерфейс для домашних питомцев (наследование интерфейсов)
public interface IDomestic : IAnimal
{
    void Pet();
}

// 3. Базовый класс Animal
public abstract class Animal
{
    private string _name;
    private int _age;
    protected int _energy;

    public string Name
    {
        get => _name;
        set => _name = string.IsNullOrWhiteSpace(value) ? "Без клички" : value;
    }

    public int Age
    {
        get => _age;
        set
        {
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
        }
    }

    public Animal(string name, int age, int initialEnergy)
    {
        Name = name;
        Age = age;
        _energy = initialEnergy > 0 ? initialEnergy : 10;
    }

    public virtual string GetInfo()
    {
        return $"Животное: {Name} | Возраст: {Age} лет(года)";
    }
}

// 4. Класс Dog с явной реализацией IDomestic и IAnimal
public class Dog : Animal, IDomestic
{
    private string _breed;

    public string Breed
    {
        get => _breed;
        set => _breed = value;
    }

    public Dog(string name, int age, int energy, string breed) : base(name, age, energy)
    {
        Breed = breed;
    }

    // Явная реализация IAnimal
    void IAnimal.MakeSound()
    {
        Console.WriteLine($"[Явная реализация IAnimal] Собака {Name} (Порода: {Breed}) лает: Гав-гав!");
    }

    void IAnimal.Sleep(int hours)
    {
        _energy += hours * 10;
        Console.WriteLine($"[Явная реализация IAnimal] Пес {Name} поспал {hours} ч. Энергия: {_energy}");
    }

    int IAnimal.GetEnergy()
    {
        return _energy;
    }

    // Явная реализация IDomestic
    void IDomestic.Pet()
    {
        Console.WriteLine($"[Явная реализация IDomestic] Вы погладили пса {Name}. Он радостно завилял хвостом.");
    }

    // Обычный собственный метод класса
    public void Fetch()
    {
        Console.WriteLine($"Пес {Name} побежал за палкой!");
    }
}

// 5. Класс Cat с явной реализацией IDomestic и IAnimal
public class Cat : Animal, IDomestic
{
    private bool _isIndoor;

    public bool IsIndoor
    {
        get => _isIndoor;
        set => _isIndoor = value;
    }

    public Cat(string name, int age, int energy, bool isIndoor) : base(name, age, energy)
    {
        IsIndoor = isIndoor;
    }

    // Явная реализация методов
    void IAnimal.MakeSound()
    {
        Console.WriteLine($"[Явная реализация IAnimal] Кошка {Name} нежно мурлычет: Мяу-мяу~");
    }

    void IAnimal.Sleep(int hours)
    {
        _energy += hours * 8;
        Console.WriteLine($"[Явная реализация IAnimal] Кошка {Name} сладко поспала на диване {hours} ч. Энергия: {_energy}");
    }

    int IAnimal.GetEnergy()
    {
        return _energy;
    }

    void IDomestic.Pet()
    {
        Console.WriteLine($"[Явная реализация IDomestic] Вы погладили кошку {Name}. Она замурлыкала.");
    }
}

// 6. Класс Bird с явной реализацией IAnimal
public class Bird : Animal, IAnimal
{
    private double _wingspan;

    public double Wingspan
    {
        get => _wingspan;
        set
        {
            if (value > 0)
                _wingspan = value;
            else
                throw new ArgumentOutOfRangeException("Размах крыльев должен быть положительным!");
        }
    }

    public Bird(string name, int age, int energy, double wingspan) : base(name, age, energy)
    {
        Wingspan = wingspan;
    }

    // Явная реализация IAnimal
    void IAnimal.MakeSound()
    {
        Console.WriteLine($"[Явная реализация IAnimal] Птица {Name} (Размах: {Wingspan} м) поет: Чик-чирик!");
    }

    void IAnimal.Sleep(int hours)
    {
        _energy += hours * 5;
        Console.WriteLine($"[Явная реализация IAnimal] Птица {Name} поспала в гнезде {hours} ч. Энергия: {_energy}");
    }

    int IAnimal.GetEnergy()
    {
        return _energy;
    }

    public void Fly()
    {
        Console.WriteLine($"Птица {Name} расправила крылья и взмыла ввысь.");
    }
}

// --- Демонстрация работы ---
Console.WriteLine("=== 1. Обращение через ссылку на класс vs интерфейс ===");
Dog myDog = new Dog("Бобик", 3, 40, "Овчарка");
Console.WriteLine(myDog.GetInfo());
myDog.Fetch(); // Доступен собственный метод

// myDog.MakeSound(); // Ошибка компиляции, метод скрыт из публичного API класса!
// Вызов через приведение к интерфейсу:
IDomestic domesticDog = myDog;
domesticDog.MakeSound();
domesticDog.Pet();
domesticDog.Sleep(2);
Console.WriteLine($"Энергия собаки через интерфейс: {domesticDog.GetEnergy()}");

Console.WriteLine("\n=== 2. Полиморфный список с явной реализацией интерфейса ===");
List<IAnimal> animals = new List<IAnimal>
{
    new Dog("Рекс", 4, 50, "Хаски"),
    new Cat("Белла", 2, 35, true),
    new Bird("Кеша", 1, 20, 0.25)
};

foreach (IAnimal animal in animals)
{
    animal.MakeSound();
    animal.Sleep(1);
    Console.WriteLine($"Текущий уровень энергии: {animal.GetEnergy()}\n");
}
````
