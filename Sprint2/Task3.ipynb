### Вариант задания № 6




<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Movie в C#, который будет представлять информацию о
фильмах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Movie:
• Атрибуты: Название (Title), Режиссер (Director), Год выпуска (ReleaseYear).
• Методы:
o GetInfo(): метод для получения информации о фильме в виде строки.
o Watch(): метод для вывода сообщения о просмотре фильма.
o Rate(): метод для оценки фильма.

Требования к производным классам:
1. Документальный фильм (Documentary): Должен содержать дополнительные
атрибуты, такие как Тематика (Theme). Метод Watch() должен быть
переопределен для добавления информации о тематике при просмотре
документального фильма.
2. Игровой фильм (FeatureFilm): Должен содержать дополнительные атрибуты,
такие как Жанр (Genre). Метод Rate() должен быть переопределен для
добавления логики оценки в зависимости от жанра.
3. Анимационный фильм (AnimatedMovie) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Студия анимации
(AnimationStudio). Метод GetInfo() должен быть переопределен для
включения информации о студии анимации в описании фильма.



#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```
public interface IStreamable
{
    string StreamingPlatform { get; set; }
    void StreamOnline();
}

public interface IAwardWinning
{
    string AwardName { get; set; }
    void ShowAwards();
}

public class Movie : IStreamable, IAwardWinning
{
    public string Title { get; set; }
    public string Director { get; set; }
    public int ReleaseYear { get; set; }
    
    public int DurationMinutes { get; set; } 
    public decimal Budget { get; set; }      
    public string Country { get; set; }

    public string StreamingPlatform { get; set; }
    public string AwardName { get; set; }

    public Movie(string title, string director, int releaseYear, int duration, decimal budget, string country, string platform, string award)
    {
        Title = title;
        Director = director;
        ReleaseYear = releaseYear;
        DurationMinutes = duration;
        Budget = budget;
        Country = country;
        StreamingPlatform = platform;
        AwardName = award;
    }

    public virtual string GetInfo()
    {
        return $"Фильм: {Title} | Режисер: {Director} | Год: {ReleaseYear} | Длительность: {DurationMinutes} мин. | Страна: {Country}";
    }

    public virtual void Watch()
    {
        Console.WriteLine($"Вы начали просмотр фильма «{Title}».");
    }

    public virtual void Rate(double score)
    {
        Console.WriteLine($"Оценка фильма «{Title}»: {score}/10");
    }

    public void CalculateRevenue(decimal boxOffice)
    {
        decimal profit = boxOffice - Budget;
        Console.WriteLine($"Финансовый результат фильма «{Title}»: прибыль составила {profit} млн.$");
    }

    public void StreamOnline()
    {
        Console.WriteLine($"Фильм «{Title}» доступен для онлайн-просмотра на платформе {StreamingPlatform}.");
    }

    public void ShowAwards()
    {
        Console.WriteLine($"Фильм «{Title}» имеет награду: {AwardName}.");
    }

    public void DisplayProductionInfo()
    {
        Console.WriteLine($"Производство фильма «{Title}» обошлось в {Budget} млн.$ ({Country}).");
    }
}

public class Documentary : Movie
{
    public string Theme { get; set; }
    public bool HasRealFootage { get; set; }

    public Documentary(string title, string director, int releaseYear, int duration, decimal budget, string country, string platform, string award, string theme, bool hasRealFootage) 
        : base(title, director, releaseYear, duration, budget, country, platform, award)
    {
        Theme = theme;
        HasRealFootage = hasRealFootage;
    }

    public override void Watch()
    {
        Console.WriteLine($"Вы смотрите документальный фильм «{Title}» на тему «{Theme}» (Реальная хроника: {HasRealFootage}).");
    }

    public void ShowEducationalValue()
    {
        Console.WriteLine($"Документальный фильм «{Title}» рекомендован для образовательных целей в области: {Theme}.");
    }
}

public class FeatureFilm : Movie
{
    public string Genre { get; set; }
    public string AgeRating { get; set; }

    public FeatureFilm(string title, string director, int releaseYear, int duration, decimal budget, string country, string platform, string award, string genre, string ageRating) 
        : base(title, director, releaseYear, duration, budget, country, platform, award)
    {
        Genre = genre;
        AgeRating = ageRating;
    }

    public override void Rate(double score)
    {
        if (Genre == "Драма" && score >= 8.5)
        {
            Console.WriteLine($"Игровой фильм «{Title}» получил статус «Выбор критиков» с оценкой {score}/10!");
        }
        else
        {
            Console.WriteLine($"Оценка игрового фильма «{Title}» ({Genre}, рейтинг {AgeRating}): {score}/10.");
        }
    }

    public void DisplayBoxOfficeInfo(decimal gross)
    {
        Console.WriteLine($"Кассовые сборы фильма «{Title}» составили {gross} млн.$. Возрастное ограничение: {AgeRating}.");
    }
}

public class AnimatedMovie : Movie
{
    public string AnimationStudio { get; set; }
    public string AnimationTechnique { get; set; }

    public AnimatedMovie(string title, string director, int releaseYear, int duration, decimal budget, string country, string platform, string award, string animationStudio, string technique) 
        : base(title, director, releaseYear, duration, budget, country, platform, award)
    {
        AnimationStudio = animationStudio;
        AnimationTechnique = technique;
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()} | Студия: {AnimationStudio} | Техника: {AnimationTechnique}";
    }

    public void PlayThemeSong()
    {
        Console.WriteLine($"В анимационном фильме «{Title}» играет фирменная заглавная песня студии {AnimationStudio}.");
    }
}

public class Series : Movie
{
    public int SeasonsCount { get; set; }

    public Series(string title, string director, int releaseYear, int duration, decimal budget, string country, string platform, string award, int seasons) 
        : base(title, director, releaseYear, duration, budget, country, platform, award)
    {
        SeasonsCount = seasons;
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()} | Всего сезонов: {SeasonsCount}";
    }
}

public class TvShow : Series
{
    public string Channel { get; set; }

    public TvShow(string title, string director, int releaseYear, int duration, decimal budget, string country, string platform, string award, int seasons, string channel) 
        : base(title, director, releaseYear, duration, budget, country, platform, award, seasons)
    {
        Channel = channel;
    }

    public override void Watch()
    {
        Console.WriteLine($"Вы смотрите ТВ-шоу «{Title}» на канале {Channel} (Сезонов: {SeasonsCount}).");
    }
}

Documentary doc = new Documentary("Планета Земля", "Аластер Фозергилл", 2006, 500, 25m, "Великобритания", "Netflix", "Эмми", "Дикая природа", true);
FeatureFilm feature = new FeatureFilm("Оппенгеймер", "Кристофер Нолан", 2023, 180, 100m, "США", "Peacock", "Оскар", "Драма", "18+");
AnimatedMovie animated = new AnimatedMovie("Унесённые призраками", "Хаяо Миядзаки", 2001, 125, 19m, "Япония", "Max", "Оскар", "Studio Ghibli", "Рисованная анимация");
TvShow tvShow = new TvShow("Наша планета", "Дэвид Аттенборо", 2019, 400, 30m, "Великобритания", "Netflix", "BAFTA", 2, "BBC");

Console.WriteLine("=== 1. ДОКУМЕНТАЛЬНЫЙ ФИЛЬМ (Простое наследование + Интерфейсы) ===");
Console.WriteLine(doc.GetInfo());
doc.Watch();
doc.ShowEducationalValue();
doc.StreamOnline();
doc.ShowAwards();
doc.CalculateRevenue(60m);

Console.WriteLine("\n=== 2. ИГРОВОЙ ФИЛЬМ (Полиморфизм) ===");
Console.WriteLine(feature.GetInfo());
feature.Rate(9.0);
feature.DisplayBoxOfficeInfo(950m);
feature.StreamOnline();

Console.WriteLine("\n=== 3. АНИМАЦИОННЫЙ ФИЛЬМ ===");
Console.WriteLine(animated.GetInfo());
animated.PlayThemeSong();
animated.ShowAwards();

Console.WriteLine("\n=== 4. ТВ-ШОУ (Сложное/Многоуровневое наследование: Movie -> Series -> TvShow) ===");
Console.WriteLine(tvShow.GetInfo());
tvShow.Watch();
```
