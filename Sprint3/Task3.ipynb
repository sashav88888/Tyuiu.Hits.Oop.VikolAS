### Вариант задания № 6

Создать базовый класс Movie в C#, который будет представлять информацию о фильмах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Movie:
• Атрибуты: Название (Title), Режиссер (Director), Год выпуска (ReleaseYear).
• Методы:
  o GetInfo(): метод для получения информации о фильме в виде строки.
  o Watch(): метод для вывода сообщения о просмотре фильма.
  o Rate(): метод для оценки фильма.

Требования к производным классам:
1. Документальный фильм (Documentary): дополнительные атрибуты (Theme). Метод Watch() переопределен для вывода тематики.
2. Игровой фильм (FeatureFilm): дополнительные атрибуты (Genre). Метод Rate() переопределен с добавлением логики в зависимости от жанра.
3. Анимационный фильм (AnimatedMovie): дополнительные атрибуты (AnimationStudio). Метод GetInfo() переопределен для включения студии анимации.

```
using System;
using System.Collections.Generic;
using System.Linq;

// ==========================================
// 1. БАЗОВЫЙ КЛАСС MOVIE
// ==========================================
public class Movie
{
    // Приватные поля (инкапсуляция)
    private string _title;
    private string _director;
    private int _releaseYear;
    private int _durationMinutes;
    private decimal _budget;
    private string _country;

    // Свойства с валидацией
    public string Title
    {
        get => _title;
        set => _title = string.IsNullOrWhiteSpace(value) ? "Без названия" : value;
    }

    public string Director
    {
        get => _director;
        set => _director = string.IsNullOrWhiteSpace(value) ? "Неизвестный режиссер" : value;
    }

    public int ReleaseYear
    {
        get => _releaseYear;
        set
        {
            if (value > 1888)
                _releaseYear = value;
            else
                throw new ArgumentOutOfRangeException("Год выпуска должен быть позже 1888 года!");
        }
    }

    public int DurationMinutes
    {
        get => _durationMinutes;
        set
        {
            if (value > 0)
                _durationMinutes = value;
            else
                throw new ArgumentOutOfRangeException("Длительность фильма должна быть больше 0 минут!");
        }
    }

    public decimal Budget
    {
        get => _budget;
        set
        {
            if (value >= 0)
                _budget = value;
            else
                throw new ArgumentOutOfRangeException("Бюджет не может быть отрицательным!");
        }
    }

    public string Country
    {
        get => _country;
        set => _country = string.IsNullOrWhiteSpace(value) ? "Не указана" : value;
    }

    // Статическое свойство
    public static int TotalMovies { get; private set; } = 0;

    // Конструктор
    public Movie(string title, string director, int releaseYear, int duration, decimal budget, string country)
    {
        Title = title;
        Director = director;
        ReleaseYear = releaseYear;
        DurationMinutes = duration;
        Budget = budget;
        Country = country;
        TotalMovies++;
    }

    // --- Виртуальные методы для полиморфного перекрытия (override) ---
    public virtual string GetInfo()
    {
        return $"Фильм: «{Title}» | Режиссер: {Director} | Год: {ReleaseYear} | Хронометраж: {DurationMinutes} мин. | Страна: {Country}";
    }

    public virtual void Watch()
    {
        Console.WriteLine($"[Просмотр] Вы начали просмотр фильма «{Title}».");
    }

    public virtual void Rate(double score)
    {
        Console.WriteLine($"[Оценка] Фильм «{Title}» оценен на {score}/10.");
    }

    // --- Перегрузка методов (Overload) ---
    // Перегрузка Watch: указание качества видео
    public void Watch(string resolution)
    {
        Console.WriteLine($"[Просмотр] Воспроизведение фильма «{Title}» в разрешении {resolution}.");
    }

    // Перегрузка Watch: указание таймкода продолжения просмотра
    public void Watch(int resumeFromMinute)
    {
        Console.WriteLine($"[Просмотр] Продолжение фильма «{Title}» с {resumeFromMinute}-й минуты.");
    }

    // Перегрузка Rate: оценка с текстовым отзывом
    public void Rate(double score, string review)
    {
        Console.WriteLine($"[Оценка и Рецензия] Фильм «{Title}» ({score}/10). Рецензия: \"{review}\"");
    }

    // Дополнительные методы базового класса
    public void CalculateProfit(decimal boxOffice)
    {
        decimal profit = boxOffice - Budget;
        Console.WriteLine($"[Экономика] «{Title}»: Бюджет {Budget} млн.$, Сборы {boxOffice} млн.$, Прибыль: {profit} млн.$");
    }

    public void DisplayAgeOfMovie(int currentYear = 2026)
    {
        Console.WriteLine($"[Возраст картины] С момента премьеры «{Title}» прошло {currentYear - ReleaseYear} лет.");
    }

    public static void DisplayTotalMovies()
    {
        Console.WriteLine($"\n[Статистика] Всего создано фильмов в каталоге: {TotalMovies}");
    }
}

// ==========================================
// 2. ПРОИЗВОДНЫЙ КЛАСС DOCUMENTARY
// ==========================================
public class Documentary : Movie
{
    private string _theme;
    public string Theme
    {
        get => _theme;
        set => _theme = value;
    }

    public bool HasRealFootage { get; set; }
    public string ResearchSource { get; set; }
    public string Narrator { get; set; }

    public Documentary(string title, string director, int releaseYear, int duration, decimal budget, string country,
                       string theme, bool hasRealFootage, string researchSource, string narrator)
        : base(title, director, releaseYear, duration, budget, country)
    {
        Theme = theme;
        HasRealFootage = hasRealFootage;
        ResearchSource = researchSource;
        Narrator = narrator;
    }

    // Перекрытие метода Watch()
    public override void Watch()
    {
        Console.WriteLine($"[Просмотр документалки] Вы смотрите «{Title}» на тему «{Theme}» (Диктор: {Narrator}).");
    }

    // Перегрузка метода Watch() для документального кино
    public void Watch(bool showSubtitles)
    {
        string subStatus = showSubtitles ? "включены" : "выключены";
        Console.WriteLine($"[Просмотр документалки] «{Title}» на тему «{Theme}». Научные субтитры: {subStatus}.");
    }

    public void ShowFactSheet()
    {
        Console.WriteLine($"[Факты] Фильм «{Title}» основан на источнике: «{ResearchSource}». Реальные архивные кадры: {(HasRealFootage ? "Да" : "Нет")}.");
    }

    public void RecommendForStudy(string subject)
    {
        Console.WriteLine($"[Образование] Фильм «{Title}» официально рекомендован кафедрой: {subject}.");
    }
}

// ==========================================
// 3. ПРОИЗВОДНЫЙ КЛАСС FEATUREFILM
// ==========================================
public class FeatureFilm : Movie
{
    private string _genre;
    public string Genre
    {
        get => _genre;
        set => _genre = value;
    }

    public string AgeRating { get; set; }
    public string MainActor { get; set; }
    public int NominationsCount { get; set; }

    public FeatureFilm(string title, string director, int releaseYear, int duration, decimal budget, string country,
                       string genre, string ageRating, string mainActor, int nominations)
        : base(title, director, releaseYear, duration, budget, country)
    {
        Genre = genre;
        AgeRating = ageRating;
        MainActor = mainActor;
        NominationsCount = nominations;
    }

    // Перекрытие метода Rate() с жанровой логикой
    public override void Rate(double score)
    {
        if (Genre == "Драма" && score >= 8.5)
        {
            Console.WriteLine($"[Выбор критиков] Игровая драма «{Title}» получила наивысший знак качества с оценкой {score}/10!");
        }
        else
        {
            Console.WriteLine($"[Оценка] Игровой фильм «{Title}» (Жанр: {Genre}, Рейтинг: {AgeRating}) оценен на {score}/10.");
        }
    }

    // Перегрузка Rate: учет зрительских симпатий
    public void Rate(double criticScore, double audienceScore)
    {
        double average = (criticScore + audienceScore) / 2.0;
        Console.WriteLine($"[Сводный рейтинг] «{Title}»: Критики - {criticScore}, Зрители - {audienceScore}. Итог: {average:F1}/10.");
    }

    public void DisplayCast()
    {
        Console.WriteLine($"[Кастинг] Главную роль в «{Title}» исполнил: {MainActor}.");
    }

    public void PrintAwardsPotential()
    {
        Console.WriteLine($"[Кинопремии] Фильм выдвинут в {NominationsCount} номинациях киноакадемии.");
    }
}

// ==========================================
// 4. ПРОИЗВОДНЫЙ КЛАСС ANIMATEDMOVIE
// ==========================================
public class AnimatedMovie : Movie
{
    private string _animationStudio;
    public string AnimationStudio
    {
        get => _animationStudio;
        set => _animationStudio = value;
    }

    public string AnimationTechnique { get; set; }
    public string TargetAudience { get; set; }
    public bool Has3DVersion { get; set; }

    public AnimatedMovie(string title, string director, int releaseYear, int duration, decimal budget, string country,
                         string studio, string technique, string targetAudience, bool has3D)
        : base(title, director, releaseYear, duration, budget, country)
    {
        AnimationStudio = studio;
        AnimationTechnique = technique;
        TargetAudience = targetAudience;
        Has3DVersion = has3D;
    }

    // Перекрытие метода GetInfo() с вызовом base.GetInfo()
    public override string GetInfo()
    {
        return $"{base.GetInfo()} | Студия анимации: {AnimationStudio} | Техника: {AnimationTechnique}";
    }

    public void PlaySoundtrack()
    {
        Console.WriteLine($"[Саундтрек] Играет титульная анимационная тема студии {AnimationStudio}.");
    }

    public void CheckCinemaFormat()
    {
        Console.WriteLine($"[Формат кинопроката] «{Title}» (Аудитория: {TargetAudience}). Поддержка 3D-очков: {(Has3DVersion ? "Есть" : "Нет")}.");
    }
}

// ==========================================
// 5. GENERIC КЛАСС ДЛЯ УПРАВЛЕНИЯ ФИЛЬМАМИ
// ==========================================
public class MovieLibrary<T> where T : Movie
{
    private List<T> _collection = new List<T>();
    public string LibraryName { get; set; }

    public MovieLibrary(string name)
    {
        LibraryName = name;
    }

    public void AddMovie(T movie)
    {
        _collection.Add(movie);
        Console.WriteLine($"[Медиатека «{LibraryName}»] Добавлен фильм: «{movie.Title}»");
    }

    public void RemoveMovie(T movie)
    {
        if (_collection.Remove(movie))
            Console.WriteLine($"[Медиатека «{LibraryName}»] Фильм «{movie.Title}» удален из каталога.");
    }

    public void ShowAllInfo()
    {
        Console.WriteLine($"\n================== КАТАЛОГ МЕДИАТЕКИ: {LibraryName.ToUpper()} ==================");
        foreach (var movie in _collection)
        {
            Console.WriteLine(movie.GetInfo());
        }
    }

    public void WatchAll()
    {
        Console.WriteLine($"\n--- Полиморфный запуск фильмов из каталога «{LibraryName}» ---");
        foreach (var movie in _collection)
        {
            movie.Watch();
        }
    }

    public int MoviesCount => _collection.Count;
}

// ==========================================
// ДЕМОНСТРАЦИЯ РАБОТЫ ВСЕХ КОМПОНЕНТОВ
// ==========================================

// Создание объектов
Documentary doc = new Documentary(
    "Планета Земля", "Аластер Фозергилл", 2006, 500, 25m, "Великобритания",
    "Дикая природа", true, "Архивы BBC Earth", "Дэвид Аттенборо"
);

FeatureFilm drama = new FeatureFilm(
    "Оппенгеймер", "Кристофер Нолан", 2023, 180, 100m, "США",
    "Драма", "18+", "Киллиан Мерфи", 13
);

AnimatedMovie anime = new AnimatedMovie(
    "Унесённые призраками", "Хаяо Миядзаки", 2001, 125, 19m, "Япония",
    "Studio Ghibli", "Традиционная рисованная", "Для всей семьи", false
);

// 1. Полиморфизм и перекрытие методов (override)
Console.WriteLine("=== 1. ПОЛИМОРФИЗМ И ПЕРЕКРЫТИЕ (OVERRIDE) ===");
Console.WriteLine(doc.GetInfo());
doc.Watch();
doc.Rate(9.4);

Console.WriteLine();
Console.WriteLine(drama.GetInfo());
drama.Watch();
drama.Rate(9.2); // Сработает логика драмы с оценкой >= 8.5

Console.WriteLine();
Console.WriteLine(anime.GetInfo()); // Выведет базовую инфу + студию анимации
anime.Watch();
anime.Rate(8.9);

// 2. Демонстрация перегрузок методов (Overload)
Console.WriteLine("\n=== 2. ДЕМОНСТРАЦИЯ ПЕРЕГРУЗКИ МЕТОДОВ (OVERLOAD) ===");
doc.Watch("4K Ultra HD");
drama.Watch(45);
doc.Watch(true);
drama.Rate(9.3, "Глубокая психологическая работа и шедевральный саундтрек.");
drama.Rate(9.5, 9.1);

// 3. Дополнительные атрибуты и методы
Console.WriteLine("\n=== 3. ДОПОЛНИТЕЛЬНЫЙ ФУНКЦИОНАЛ КЛАССОВ ===");
doc.ShowFactSheet();
doc.RecommendForStudy("Биология и Экология");
drama.CalculateProfit(957m);
drama.DisplayCast();
anime.PlaySoundtrack();
anime.CheckCinemaFormat();

// 4. Демонстрация работы Generic-классов
Console.WriteLine("\n=== 4. GENERIC КЛАССЫ (MovieLibrary<T>) ===");

// А) Общая медиатека фильмов базового типа Movie (полиморфная коллекция)
MovieLibrary<Movie> cinemaClub = new MovieLibrary<Movie>("Киноклуб ВГИК");
cinemaClub.AddMovie(doc);
cinemaClub.AddMovie(drama);
cinemaClub.AddMovie(anime);

cinemaClub.ShowAllInfo();
cinemaClub.WatchAll();

// Б) Специализированная типизированная медиатека исключительно для аниме/мультфильмов
Console.WriteLine();
MovieLibrary<AnimatedMovie> animationVault = new MovieLibrary<AnimatedMovie>("Архив Анимации");
animationVault.AddMovie(anime);
animationVault.ShowAllInfo();

// Итоговая статистика
Movie.DisplayTotalMovies();
```
